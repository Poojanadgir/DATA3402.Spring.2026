1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class Counter:
    def __init__(self, max_value):
        self.max_value = max_value
        self.value = 0

    def increment(self):
        if self.value < self.max_value:
            self.value += 1
        else:
            print("Error: Counter has reached its maximum value.")

    def reset(self):
        self.value = 0

    def get_value(self):
        return self.value
```

```python
counter = Counter(3)

counter.increment()
print(counter.get_value())  

counter.increment()
counter.increment()
print(counter.get_value())  

counter.increment()  

counter.reset()
print(counter.get_value())  
```

```python
import math

class Triangle(Shape):
    def __init__(self, side1, side2, side3, x, y):
        super().__init__(x, y)
        self.__side1 = side1
        self.__side2 = side2
        self.__side3 = side3

    # Accessors
    def get_side1(self):
        return self.__side1

    def get_side2(self):
        return self.__side2

    def get_side3(self):
        return self.__side3

    def area(self):
        s = (self.__side1 + self.__side2 + self.__side3) / 2
        return math.sqrt(
            s *
            (s - self.__side1) *
            (s - self.__side2) *
            (s - self.__side3)
        )

    def perimeter(self):
        return self.__side1 + self.__side2 + self.__side3
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the parameter of the object. 

```python
class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def area(self):
        raise NotImplementedError("Subclass must implement area method")

    def perimeter(self):
        raise NotImplementedError("Subclass must implement perimeter method")

    def get_perimeter_points(self, num_points=16):
        raise NotImplementedError("Subclass must implement get_perimeter_points")
```

```python
class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_perimeter_points(self, num_points=16):
        points = []
        l = self.__length
        w = self.__width

        perim = self.perimeter()

        for i in range(num_points):
            d = (i / num_points) * perim

            if d < l:
                points.append((self._x + d, self._y))
            elif d < l + w:
                points.append((self._x + l, self._y + (d - l)))
            elif d < 2*l + w:
                points.append((self._x + l - (d - (l + w)), self._y + w))
            else:
                points.append((self._x, self._y + w - (d - (2*l + w))))

        return points
```

```python
import math

class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_perimeter_points(self, num_points=16):
        points = []
        for i in range(num_points):
            angle = 2 * math.pi * i / num_points
            px = self._x + self.__radius * math.cos(angle)
            py = self._y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points
```

```python
class Triangle(Shape):
    def __init__(self, side1, side2, side3, x, y):
        super().__init__(x, y)
        self.__side1 = side1
        self.__side2 = side2
        self.__side3 = side3

        self._p1 = (x, y)
        self._p2 = (x + side1, y)
        self._p3 = (x + side2/2, y + side3)

    def area(self):
        s = (self.__side1 + self.__side2 + self.__side3) / 2
        return math.sqrt(s*(s-self.__side1)*(s-self.__side2)*(s-self.__side3))

    def perimeter(self):
        return self.__side1 + self.__side2 + self.__side3

    def get_perimeter_points(self, num_points=16):
        points = []
        vertices = [self._p1, self._p2, self._p3]
        edges = [(vertices[i], vertices[(i+1)%3]) for i in range(3)]

        for i in range(num_points):
            edge_index = i % 3
            t = (i // 3) / (num_points // 3)

            x = edges[edge_index][0][0] + t * (edges[edge_index][1][0] - edges[edge_index][0][0])
            y = edges[edge_index][0][1] + t * (edges[edge_index][1][1] - edges[edge_index][0][1])
            points.append((x, y))

        return points
```

```python
shapes = [
    Rectangle(5, 3, 0, 0),
    Circle(4, 1, 2),
    Triangle(3, 4, 5, 0, 0)
]

for shape in shapes:
    print(shape.get_perimeter_points())
```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    def area(self):
        raise NotImplementedError

    def perimeter(self):
        raise NotImplementedError

    def get_perimeter_points(self, num_points=16):
        raise NotImplementedError

    def contains_point(self, x, y):
        raise NotImplementedError("Subclass must implement contains_point")
```

```python
class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def contains_point(self, x, y):
        return (
            self._x <= x <= self._x + self.__length and
            self._y <= y <= self._y + self.__width
        )
```

```python
import math

class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def contains_point(self, x, y):
        return (
            (x - self._x) ** 2 +
            (y - self._y) ** 2
        ) <= self.__radius ** 2
```

```python
class Triangle(Shape):
    def __init__(self, side1, side2, side3, x, y):
        super().__init__(x, y)
        self.__side1 = side1
        self.__side2 = side2
        self.__side3 = side3

        # Define vertices (simple layout)
        self._p1 = (x, y)
        self._p2 = (x + side1, y)
        self._p3 = (x + side2/2, y + side3)

    def area(self):
        s = (self.__side1 + self.__side2 + self.__side3) / 2
        return math.sqrt(s*(s-self.__side1)*(s-self.__side2)*(s-self.__side3))

    def perimeter(self):
        return self.__side1 + self.__side2 + self.__side3

    def __triangle_area(self, p1, p2, p3):
        return abs(
            p1[0]*(p2[1]-p3[1]) +
            p2[0]*(p3[1]-p1[1]) +
            p3[0]*(p1[1]-p2[1])
        ) / 2

    def contains_point(self, x, y):
        p = (x, y)

        A = self.__triangle_area(self._p1, self._p2, self._p3)
        A1 = self.__triangle_area(p, self._p2, self._p3)
        A2 = self.__triangle_area(self._p1, p, self._p3)
        A3 = self.__triangle_area(self._p1, self._p2, p)

        return abs(A - (A1 + A2 + A3)) < 1e-6
```

```python
shapes = [
    Rectangle(5, 3, 0, 0),
    Circle(4, 1, 2),
    Triangle(3, 4, 5, 0, 0)
]

for shape in shapes:
    print(shape.contains_point(2, 1))
```

9. Add a function in the base class of the object classes that returns true/false testing that the object overlaps with another object.

```python
class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    def area(self):
        raise NotImplementedError

    def perimeter(self):
        raise NotImplementedError

    def get_perimeter_points(self, num_points=16):
        raise NotImplementedError

    def contains_point(self, x, y):
        raise NotImplementedError

    def overlaps(self, other):
    
        for (px, py) in self.get_perimeter_points():
            if other.contains_point(px, py):
                return True

        for (px, py) in other.get_perimeter_points():
            if self.contains_point(px, py):
                return True

        return False
```

```python
rect = Rectangle(5, 3, 0, 0)
circle = Circle(4, 3, 2)
triangle = Triangle(3, 4, 5, 10, 10)

print(rect.overlaps(circle))      
print(rect.overlaps(triangle))    
```

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b
        
    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"
    
    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()
        
   
def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```
