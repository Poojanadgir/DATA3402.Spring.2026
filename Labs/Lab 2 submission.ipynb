*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def check_winner(board):
    n = len(board)

    for i in range(n):
        if board[i][0] != 0:
            win = True
            for j in range(n):
                if board[i][j] != board[i][0]:
                    win = False
            if win:
                return board[i][0]

    for j in range(n):
        if board[0][j] != 0:
            win = True
            for i in range(n):
                if board[i][j] != board[0][j]:
                    win = False
            if win:
                return board[0][j]

    if board[0][0] != 0:
        win = True
        for i in range(n):
            if board[i][i] != board[0][0]:
                win = False
        if win:
            return board[0][0]

    if board[0][n-1] != 0:
        win = True
        for i in range(n):
            if board[i][n-1-i] != board[0][n-1]:
                win = False
        if win:
            return board[0][n-1]

    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                return -1

    return 0
````

````python
winner_is_2 = [[2, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_1 = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_also_1 = [[0, 1, 0],
	[2, 1, 0],
	[2, 1, 1]]

no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 2]]

also_no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 0]]


print(check_winner(winner_is_2))        
print(check_winner(winner_is_1))        
print(check_winner(winner_is_also_1))   
print(check_winner(no_winner))          
print(check_winner(also_no_winner))     
````

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def place_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def draw_tic_tac_toe_labeled(board):
    print("   A   B   C")
    for i in range(3):
        print("  --- --- ---")
        print(i + 1, end=" ")
        for j in range(3):
            if board[i][j] == 0:
                print("|   ", end="")
            elif board[i][j] == 1:
                print("| X ", end="")
            elif board[i][j] == 2:
                print("| O ", end="")
        print("|")
    print("  --- --- ---")


# PLACE MOVE USING ROW & COL
def place_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False


# PLACE MOVE
def place_move_location(board, player, location):
    if location[0] == "A":
        col = 0
    elif location[0] == "B":
        col = 1
    elif location[0] == "C":
        col = 2

    row = int(location[1]) - 1
    return place_move(board, player, row, col)


#PLAYER INPUT
def player_turn(board, player):
    valid_move = False
    while valid_move == False:
        location = input("Enter a location (A1, B2, C3): ")
        valid_move = place_move_location(board, player, location)


def check_winner(board):
    for i in range(3):
        if board[i][0] != 0 and board[i][0] == board[i][1] == board[i][2]:
            return board[i][0]

    for j in range(3):
        if board[0][j] != 0 and board[0][j] == board[1][j] == board[2][j]:
            return board[0][j]

    if board[0][0] != 0 and board[0][0] == board[1][1] == board[2][2]:
        return board[0][0]

    if board[0][2] != 0 and board[0][2] == board[1][1] == board[2][0]:
        return board[0][2]

    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return -1

    return 0


def play_tic_tac_toe():
    board = [
        [0, 0, 0],
        [0, 0, 0],
        [0, 0, 0]
    ]

    player = 1
    game_over = False

    while game_over == False:
        draw_tic_tac_toe_labeled(board)
        print("Player", player, "turn")
        player_turn(board, player)

        result = check_winner(board)

        if result == 1:
            draw_tic_tac_toe_labeled(board)
            print("Player 1 wins!")
            game_over = True

        elif result == 2:
            draw_tic_tac_toe_labeled(board)
            print("Player 2 wins!")
            game_over = True

        elif result == 0:
            draw_tic_tac_toe_labeled(board)
            print("It's a draw!")
            game_over = True

        else:
            if player == 1:
                player = 2
            else:
                player = 1
````

````python
def computer_move_basic(board):
    best_score = -10
    move = None

    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                board[i][j] = 2
                if check_winner(board) == 2:   
                    board[i][j] = 2
                    return
                elif check_winner(board) == 1: 
                    score = -1
                else:
                    score = 0
                board[i][j] = 0

                if score > best_score:
                    best_score = score
                    move = (i, j)

    if move:
        board[move[0]][move[1]] = 2
    else:
        
        for i in range(3):
            for j in range(3):
                if board[i][j] == 0:
                    board[i][j] = 2
                    return
````
